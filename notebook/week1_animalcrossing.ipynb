for = 반복문 (반복할 대상에서 하나씩 꺼내서 코드를 반복 실행해라)    

ex.   
for x in [1,2,3]:  
    print(x)  

-> x에다가 [1,2,3] 값을 하나씩 넣고 print(x)를 **반복**해라  

1   
2   
3  

으로 결과가 나오게 됨.  

기본 구조:  
for 변수 in 반복할 대상:  
    반복해서 실행할 코드
