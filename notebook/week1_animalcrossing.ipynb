## **Data Wrangling** : 데이터 정제

## 데이터 삭제 등 정제

```python
wish['Name'] = wish['Name'].str.strip().drop_duplicates()
```

- data['column'] = 요 데이터프레임에서 해당 열만 가져와 

파이썬에서 [] -> 무언가를 선택하거나 가져옴 / () -> 함수나 메서드 실행

- .str.strip() = Name 관측값 안에서 앞뒤 불필요한 공백 제거 (ex. ' Coco' -> 'Coco')
- .drop_duplicates() =  중복 제거 ; 아까 말했던 Raymond같은 사례!

=> 파이썬에서 함수 실행은 .함수 로 이으면 되는데, 얘는 작업상이어지니까 괜찮지만, 요 데이터프레임에서 5행 보여주고, 기술통계량 보여줘! 이런 분리된 작업은 따로 써주자.     
=> str만 봐도 str.strip()이여서 str 기능을 거쳐서 strip()이라는 함수를 써준 것. 그래서 결론적으로는    
. -> 앞에 대상에 붙어있는 기능이나 속성에 접근 / () -> 그중 함수를 실행   

=> 그래서 결론적으로 저 함수는 기존의 문제많은 wish 데이터프레임에서 공백제거, 중복제거하고 정제된 값을 다시 고이 name 칼럼에다 넣어준 것

```python
wish[0:10]
```

- data[시작:끝] = 해당 데이터에서 '시작'부터 '끝'까지 가져와서 보여줘(범위 지정) !!0부터 시작 

근데 이랬더니 값에서 NaN(결측값)이 뜬다. 어케 해결하지? 머가 문제?

```python
wish = wish.dropna()
wish[0:10]
```

- .dropna() = 결측값이 있는 행을 제거 
 
위에서 나온 결측값을 없앴다!   
=> 데이터를 정제하는 방법 중 하나는 **!!!중복 제거!!!**

```python
wish = wish[wish['Name'].str[0].str.isupper()]
wish[0:10]
```

- str = 문자열을 다루기 위한 접근자

**str[0]** = 아까 [숫자]이거 범위지정하는 거라 했으니까 0부터, 즉 각 문자열의 첫번째 글자만 가져오기

- .str.isupper() = 그 글자가 대문자인지 확인

이거 아까 거기서 smug cat 지우는 용이네   
=> 데이터를 정제하는 두 번째 방법은 **!!!칼럼의 속성 확인 후 속성 맞춰 제거!!!**

```python
wish = wish[wish.Name != 'Not in stock']
wish[0:10]
```

- != = 다르면 True, 같으면 False 라는 값을 배정해주는 연산자 역할

**wish[wish.Name != 'Not in stock']** = wish 데이터프레임의 Name 열에서 저것과 다르면 True
-> False 나온, 즉 빼고 싶은 값 제외한 데이터프레임을 새로 생성   
=> 데이터를 정제하는 세 번째 방법은 **!!!직접 골라 제거하기!!!**   

### => 결국 **앞서서 데이터를 선 탐색 후 방법을 택해 데이터를 정제해주자**
